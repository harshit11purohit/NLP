## Word Frequency Distribution

```python
from collections import Counter
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')

def word_frequency(text):
    words = word_tokenize(text.lower())
    return Counter(words)

text = "NLP is amazing. NLP makes machines understand text."
print("Word Frequencies:", word_frequency(text))
```

```python
def word_frequency(text):
    words = text.lower().split()
    return Counter(words)

text = "NLP is amazing. NLP makes machines understand text."
print("Word Frequencies:", word_frequency(text))
```

## BOW

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "NLP is fun and exciting",
    "Machines understand NLP and text",
    "Text processing is part of NLP"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print("Feature Names:", vectorizer.get_feature_names_out())
print("BoW Representation:")
print(X.toarray())
```

Using Numpy


```python
import numpy as np

corpus = [
    "NLP is fun and exciting",
    "Machines understand NLP and text",
    "Text processing is part of NLP"
]

tokenized = [[w for w in doc.lower().split() if w.isalnum() and len(w) >= 2] for doc in corpus]
vocab = sorted(list(set(w for doc in tokenized for w in doc)))

N = len(corpus)
V = len(vocab)
print("Feature Names:", np.array(vocab))

tf = np.zeros((N, V))
for i, doc in enumerate(tokenized):
    for w in doc:
        if w in vocab:
            tf[i, vocab.index(w)] += 1

df = np.sum(tf > 0, axis=0)
idf = np.log((1 + N) / (1 + df)) + 1

tfidf = tf * idf

norms = np.linalg.norm(tfidf, axis=1, keepdims=True)
norms[norms == 0] = 1
tfidf_norm = tfidf / norms

print("\nTF-IDF Representation:")
print(np.round(tfidf_norm, 4))
```

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "NLP is fun and exciting",
    "Machines understand NLP and text",
    "Text processing is part of NLP"
]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
feature_array = np.array(vectorizer.get_feature_names_out())

global_weights = np.sum(X.toarray(), axis=0)

importance = np.argsort(global_weights)[::-1]

keywords = feature_array[importance[:5]]

print("Top Global Keywords:", keywords)
```
