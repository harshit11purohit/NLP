LEMINIZATION


```python
import spacy
nlp=spacy.load('en_core_web_sm')
words="runner,burnt , happily , running, hulk, lovely, harshit"
doc=nlp(words)
print(doc)
```

```python
#  token.lemma_ == . It looks at the word's Part-of-Speech (POS) tag to return a grammatically accurate, real-world dictionary word
lemiatized_data=[token.lemma_ for token in doc]
print(lemiatized_data)
```

```python
lemmatized_with_pos = [(token.lemma_, token.pos_) for token in doc]
print(lemmatized_with_pos)
```

```python
for token in doc:
    print(f"'{token.text}' -> lemma: '{token.lemma_}', pos: '{token.pos_}'")
```
