```python
import re

text = "The pipeline run took 142 seconds ans minute"

# Parentheses define capturing groups
pattern = r'(\d+)\s+(\w+)'
match = re.search(pattern, text)

if match:
    print("Full Match  (group 0):", match.group(0))  # The whole chunk found
    print("Digits Only (group 1):", match.group(1))  # First parenthesis content
    print("Word Only   (group 2):", match.group(2))  # Second parenthesis content
```

```python
text = "ID: 101-Alpha, ID: 202-Beta"

# Two capturing groups inside a global search
pattern = r'(\d+)-(\w+)'
results = re.findall(pattern, text)

print(results)
```

```python
text = "ID: 101-Alpha, ID: 202-Beta"

pattern = r'(\d+)-(\w+)'
match = re.search(pattern, text)
result = (match.group(1), match.group(2))
print(result)
```
